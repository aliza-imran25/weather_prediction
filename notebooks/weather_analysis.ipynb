```python
# Create lag features

df["temp_lag_1"] = df["temperature_2m_mean"].shift(1)
df["temp_lag_2"] = df["temperature_2m_mean"].shift(2)
df["temp_lag_3"] = df["temperature_2m_mean"].shift(3)
df["temp_lag_7"] = df["temperature_2m_mean"].shift(7)

# Create 7-day rolling average
df["temp_rolling_7"] = (
    df["temperature_2m_mean"]
    .shift(1)
    .rolling(7)
    .mean()
)

df[[
    "time",
    "temperature_2m_mean",
    "temp_lag_1",
    "temp_lag_2",
    "temp_lag_3",
    "temp_lag_7",
    "temp_rolling_7"
]].head(15)
```

```python
lag_features = [
    "temp_lag_1",
    "temp_lag_2",
    "temp_lag_3",
    "temp_lag_7",
    "temp_rolling_7",
    "month",
    "day_of_year"
]

# Create dataset for lag-based model
df_lag_model = df.dropna(
    subset=lag_features + ["tomorrow_temperature"]
).copy()

X_lag = df_lag_model[lag_features]
y_lag = df_lag_model["tomorrow_temperature"]

print("Rows:", len(df_lag_model))
print("Features:", X_lag.columns.tolist())
print("Target:", y_lag.name)
```

```python
# 80/20 chronological split

split_index_lag = int(len(df_lag_model) * 0.8)

X_lag_train = X_lag.iloc[:split_index_lag]
X_lag_test = X_lag.iloc[split_index_lag:]

y_lag_train = y_lag.iloc[:split_index_lag]
y_lag_test = y_lag.iloc[split_index_lag:]

print("Training samples:", len(X_lag_train))
print("Testing samples:", len(X_lag_test))

print("\nTraining period:")
print(df_lag_model["time"].iloc[0])
print("to")
print(df_lag_model["time"].iloc[split_index_lag - 1])

print("\nTesting period:")
print(df_lag_model["time"].iloc[split_index_lag])
print("to")
print(df_lag_model["time"].iloc[-1])
```

```python
from sklearn.linear_model import LinearRegression

# Create the lag-based Linear Regression model
lag_model = LinearRegression()

# Train it
lag_model.fit(X_lag_train, y_lag_train)

# Make predictions
lag_predictions = lag_model.predict(X_lag_test)

print("Lag-based Linear Regression training completed!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Evaluate lag-based model
lag_mae = mean_absolute_error(y_lag_test, lag_predictions)
lag_rmse = np.sqrt(mean_squared_error(y_lag_test, lag_predictions))
lag_r2 = r2_score(y_lag_test, lag_predictions)

print(f"Lag Model MAE: {lag_mae:.3f} °C")
print(f"Lag Model RMSE: {lag_rmse:.3f} °C")
print(f"Lag Model R²: {lag_r2:.3f}")
```

```python
# Previous-day weather features

weather_columns = [
    "temperature_2m_mean",
    "temperature_2m_max",
    "temperature_2m_min",
    "precipitation_sum",
    "rain_sum",
    "wind_speed_10m_max"
]

for column in weather_columns:
    df[f"{column}_lag_1"] = df[column].shift(1)

# Check the new features
df[[
    "time",
    "temperature_2m_mean_lag_1",
    "temperature_2m_max_lag_1",
    "temperature_2m_min_lag_1",
    "precipitation_sum_lag_1",
    "rain_sum_lag_1",
    "wind_speed_10m_max_lag_1"
]].head(10)
```

```python
weather_lag_features = [
    "temperature_2m_mean_lag_1",
    "temperature_2m_max_lag_1",
    "temperature_2m_min_lag_1",
    "precipitation_sum_lag_1",
    "rain_sum_lag_1",
    "wind_speed_10m_max_lag_1",
    "temp_lag_2",
    "temp_lag_3",
    "temp_lag_7",
    "temp_rolling_7",
    "month",
    "day_of_year"
]

df_weather_lag = df.dropna(
    subset=weather_lag_features + ["tomorrow_temperature"]
).copy()

X_weather_lag = df_weather_lag[weather_lag_features]
y_weather_lag = df_weather_lag["tomorrow_temperature"]

print("Rows:", len(df_weather_lag))
print("Number of features:", len(weather_lag_features))
print("\nFeatures:")
print(weather_lag_features)
```

```python
# 80/20 chronological split

split_index_weather = int(len(df_weather_lag) * 0.8)

X_weather_train = X_weather_lag.iloc[:split_index_weather]
X_weather_test = X_weather_lag.iloc[split_index_weather:]

y_weather_train = y_weather_lag.iloc[:split_index_weather]
y_weather_test = y_weather_lag.iloc[split_index_weather:]

print("Training samples:", len(X_weather_train))
print("Testing samples:", len(X_weather_test))

print("\nTraining period:")
print(df_weather_lag["time"].iloc[0])
print("to")
print(df_weather_lag["time"].iloc[split_index_weather - 1])

print("\nTesting period:")
print(df_weather_lag["time"].iloc[split_index_weather])
print("to")
print(df_weather_lag["time"].iloc[-1])
```

```python
from sklearn.linear_model import LinearRegression

weather_model = LinearRegression()

weather_model.fit(
    X_weather_train,
    y_weather_train
)

weather_predictions = weather_model.predict(X_weather_test)

print("Weather-History Linear Regression training completed!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

weather_mae = mean_absolute_error(
    y_weather_test,
    weather_predictions
)

weather_rmse = np.sqrt(
    mean_squared_error(
        y_weather_test,
        weather_predictions
    )
)

weather_r2 = r2_score(
    y_weather_test,
    weather_predictions
)

print(f"Weather-History Model MAE: {weather_mae:.3f} °C")
print(f"Weather-History Model RMSE: {weather_rmse:.3f} °C")
print(f"Weather-History Model R²: {weather_r2:.3f}")
```

```python
import numpy as np

df["day_of_year_sin"] = np.sin(
    2 * np.pi * df["day_of_year"] / 365
)

df["day_of_year_cos"] = np.cos(
    2 * np.pi * df["day_of_year"] / 365
)

print(
    df[
        [
            "time",
            "day_of_year",
            "day_of_year_sin",
            "day_of_year_cos"
        ]
    ].head(10)
)
```

```python
cyclical_features = [
    "temperature_2m_mean",
    "temperature_2m_max",
    "temperature_2m_min",
    "precipitation_sum",
    "rain_sum",
    "wind_speed_10m_max",
    "day_of_year_sin",
    "day_of_year_cos"
]

df_cyclical = df.dropna(
    subset=cyclical_features + ["tomorrow_temperature"]
).copy()

X_cyclical = df_cyclical[cyclical_features]
y_cyclical = df_cyclical["tomorrow_temperature"]

split_index_cyclical = int(len(df_cyclical) * 0.8)

X_cyclical_train = X_cyclical.iloc[:split_index_cyclical]
X_cyclical_test = X_cyclical.iloc[split_index_cyclical:]

y_cyclical_train = y_cyclical.iloc[:split_index_cyclical]
y_cyclical_test = y_cyclical.iloc[split_index_cyclical:]

print("Training samples:", len(X_cyclical_train))
print("Testing samples:", len(X_cyclical_test))
```

```python
from sklearn.linear_model import LinearRegression

cyclical_model = LinearRegression()

cyclical_model.fit(
    X_cyclical_train,
    y_cyclical_train
)

cyclical_predictions = cyclical_model.predict(
    X_cyclical_test
)

print("Cyclical-feature Linear Regression training completed!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

cyclical_mae = mean_absolute_error(
    y_cyclical_test,
    cyclical_predictions
)

cyclical_rmse = np.sqrt(
    mean_squared_error(
        y_cyclical_test,
        cyclical_predictions
    )
)

cyclical_r2 = r2_score(
    y_cyclical_test,
    cyclical_predictions
)

print(f"Cyclical Model MAE: {cyclical_mae:.3f} °C")
print(f"Cyclical Model RMSE: {cyclical_rmse:.3f} °C")
print(f"Cyclical Model R²: {cyclical_r2:.3f}")
```

```python
residuals = y_cyclical_test - cyclical_predictions

print("First 10 residuals:")
print(residuals.head(10).to_string(index=False))

print("\nMean residual:", residuals.mean())
print("Minimum residual:", residuals.min())
print("Maximum residual:", residuals.max())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.scatter(
    y_cyclical_test,
    residuals,
    alpha=0.5
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Actual Temperature (°C)")
plt.ylabel("Residual (°C)")
plt.title("Residual Plot - Cyclical Linear Regression")

plt.show()
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

rain_df = df.dropna(
    subset=["rain_tomorrow"]
).copy()

X_rain = rain_df[
    [
        "temperature_2m_mean",
        "temperature_2m_max",
        "temperature_2m_min",
        "precipitation_sum",
        "rain_sum",
        "wind_speed_10m_max",
        "day_of_year_sin",
        "day_of_year_cos"
    ]
]

y_rain = rain_df["rain_tomorrow"]

split_index_rain = int(len(rain_df) * 0.8)

X_rain_train = X_rain.iloc[:split_index_rain]
X_rain_test = X_rain.iloc[split_index_rain:]

y_rain_train = y_rain.iloc[:split_index_rain]
y_rain_test = y_rain.iloc[split_index_rain:]

print("Training samples:", len(X_rain_train))
print("Testing samples:", len(X_rain_test))
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

rain_baseline = DummyClassifier(
    strategy="most_frequent"
)

rain_baseline.fit(
    X_rain_train,
    y_rain_train
)

rain_baseline_predictions = rain_baseline.predict(
    X_rain_test
)

rain_baseline_accuracy = accuracy_score(
    y_rain_test,
    rain_baseline_predictions
)

print(
    f"Rain Baseline Accuracy: "
    f"{rain_baseline_accuracy:.3f}"
)
```

```python
from sklearn.linear_model import LogisticRegression

rain_model = LogisticRegression(
    max_iter=1000
)

rain_model.fit(
    X_rain_train,
    y_rain_train
)

rain_predictions = rain_model.predict(
    X_rain_test
)

print("Rain Logistic Regression training completed!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

rain_accuracy = accuracy_score(
    y_rain_test,
    rain_predictions
)

rain_precision = precision_score(
    y_rain_test,
    rain_predictions
)

rain_recall = recall_score(
    y_rain_test,
    rain_predictions
)

rain_f1 = f1_score(
    y_rain_test,
    rain_predictions
)

rain_cm = confusion_matrix(
    y_rain_test,
    rain_predictions
)

print(f"Accuracy:  {rain_accuracy:.3f}")
print(f"Precision: {rain_precision:.3f}")
print(f"Recall:    {rain_recall:.3f}")
print(f"F1 Score:  {rain_f1:.3f}")

print("\nConfusion Matrix:")
print(rain_cm)
```
